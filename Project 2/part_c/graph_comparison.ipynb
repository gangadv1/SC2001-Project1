```python
import matplotlib.pyplot as plt
import numpy as np

# Data from empirical results
v_sparse = np.array([500, 1000, 2000, 4000, 8000])
time_a_sparse = np.array([0.28, 1.08, 4.34, 16.82, 68.44])
time_b_sparse = np.array([0.18, 0.41, 0.98, 2.05, 4.54])

v_dense = np.array([200, 400, 600, 800, 1000])
time_a_dense = np.array([0.09, 0.52, 1.29, 2.25, 3.26])
time_b_dense = np.array([0.34, 1.03, 2.09, 3.34, 4.90])

plt.figure(figsize=(12, 5))

# Plot 1: Sparse Graphs Comparison
plt.subplot(1, 2, 1)
plt.plot(v_sparse, time_a_sparse, 'o-', label='(a) Matrix + Array ($O(V^2)$)', linewidth=2, color='#d95f02')
plt.plot(v_sparse, time_b_sparse, 's-', label='(b) Adj. List + Min-Heap ($O(E \\log V)$)', linewidth=2, color='#1b9e77')
plt.title('Sparse Graphs Comparison ($E = 5|V|$)', fontsize=12, fontweight='bold')
plt.xlabel('Number of Vertices $|V|$', fontsize=10)
plt.ylabel('Average Runtime (ms)', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

# Plot 2: Dense Graphs Comparison
plt.subplot(1, 2, 2)
plt.plot(v_dense, time_a_dense, 'o-', label='(a) Matrix + Array ($O(V^2)$)', linewidth=2, color='#d95f02')
plt.plot(v_dense, time_b_dense, 's-', label='(b) Adj. List + Min-Heap ($O(V^2 \\log V)$)', linewidth=2, color='#1b9e77')
plt.title('Dense Graphs Comparison ($E \\approx 0.5|V|^2$)', fontsize=12, fontweight='bold')
plt.xlabel('Number of Vertices $|V|$', fontsize=10)
plt.ylabel('Average Runtime (ms)', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.tight_layout()
plt.savefig('dijkstra_comparison.png', dpi=300)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# X-axis: Number of edges |E| at fixed |V| = 1000
edges_varying = [2000, 10000, 50000, 200000, 500000]

# Part (a): Matrix + Array
matrix_runtime = [2.5, 2.4, 2.3, 2.4, 3.4]

# Part (b): List + Min-Heap
heap_runtime = [0.249, 0.647, 1.435, 2.829, 6.457]

plt.figure(figsize=(8, 5))
plt.plot(edges_varying, matrix_runtime, marker='o', linestyle='-', linewidth=2, color='#d95f02', label='(a) matrix + array')
plt.plot(edges_varying, heap_runtime, marker='s', linestyle='-', linewidth=2, color='#1b9e77', label='(b) list + heap')


plt.xlabel('Number of Edges |E| (Fixed |V| = 1,000)')
plt.ylabel('Runtime (ms)')
plt.title('Result 1: fixed |V| = 1000, varying |E|')
plt.legend(loc='upper right')
plt.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.savefig('varying_edges_crossover_plot.png', dpi=300)
plt.show()
```
